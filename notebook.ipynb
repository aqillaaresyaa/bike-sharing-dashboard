## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.

- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

### Cleaning Data

**Fixing problem**
1. Fixing Tipe Data dteday

```python
# salin data mentah agar data asli tetap aman
day_clean = day_df.copy()
hour_clean = hour_df.copy()

# ubah dteday dari teks menjadi datetime dengan dayfirst=True
for df in (day_clean, hour_clean):
    df["dteday"] = pd.to_datetime(df["dteday"], dayfirst=True)

day_clean["dteday"].dtype
```

**Insight:** (Opsional)
- Kolom `dteday` sudah bertipe datetime pada kedua dataset

**Fixing problem**

2. Fixing Invalid Value pada hum

```python
for df in (day_clean, hour_clean):
    # nilai 0 dianggap tidak valid, ubah menjadi NaN
    df["hum"] = df["hum"].replace(0, np.nan)

    # isi NaN dengan median kelembapan pada bulan dan tahun yang sama
    df["hum"] = df["hum"].fillna(df.groupby(["yr", "mnth"])["hum"].transform("median"))

# pastikan tidak ada hum = 0 dan tidak ada NaN tersisa
print("hum = 0 :", (day_clean["hum"] == 0).sum(), (hour_clean["hum"] == 0).sum())
print("hum NaN :", day_clean["hum"].isna().sum(), hour_clean["hum"].isna().sum())
```

**Insight:** (Opsional)
- Nilai kelembapan tidak valid sudah diganti dengan median kelembapan bulan yang sama (Maret 2011)

**Fixing problem**

3. Fixing Inconsistent Value pada Kolom Kategori dan Satuan

```python
# label musim mengikuti dokumentasi dataset (kode 1 sampai 4)
season_map  = {1: "Musim Semi", 2: "Musim Panas", 3: "Musim Gugur", 4: "Musim Dingin"}
weather_map = {1: "Cerah", 2: "Berkabut", 3: "Hujan/Salju Ringan", 4: "Hujan/Salju Lebat"}
year_map    = {0: 2011, 1: 2012}
dayname_map = {0: "Minggu", 1: "Senin", 2: "Selasa", 3: "Rabu", 4: "Kamis", 5: "Jumat", 6: "Sabtu"}

for df in (day_clean, hour_clean):
    # beri label pada kolom kategori
    df["season"]     = df["season"].map(season_map)
    df["weathersit"] = df["weathersit"].map(weather_map)
    df["year"]       = df["yr"].map(year_map)
    df["weekday"]    = df["weekday"].map(dayname_map)

    # buat kolom jenis hari dari workingday
    df["day_type"]   = np.where(df["workingday"] == 1, "Hari Kerja", "Akhir Pekan/Libur")

    # kembalikan nilai ternormalisasi ke satuan asli (pengali mengikuti deskripsi dataset di Kaggle)
    df["temp_c"]     = df["temp"] * 41
    df["atemp_c"]    = df["atemp"] * 50
    df["hum_pct"]    = df["hum"] * 100
    df["wind_kmh"]   = df["windspeed"] * 67

    # rapikan nama kolom dan buang kolom lama
    df.rename(columns={"mnth": "month", "hr": "hour"}, inplace=True)
    df.drop(columns=["yr", "temp", "atemp", "hum", "windspeed"], inplace=True)

day_clean.head()
```

**Insight:** (Opsional)
- Kolom kategori sudah berlabel sehingga mudah dibaca.
- Suhu, kelembapan, dan kecepatan angin dikembalikan ke satuan asli (°C, %, km/jam) dengan pengali `temp` × 41, `atemp` × 50, `hum` × 100, dan `windspeed` × 67. Sumber pengali ini adalah deskripsi dataset di Kaggle. Berkas `Readme.txt` memakai rumus normalisasi min-maks (`temp` dengan rentang -8 sampai +39 °C dan `atemp` dengan rentang -16 sampai +50 °C), sehingga nilai suhu dapat sedikit berbeda jika memakai rumus tersebut. Seluruh kategori suhu pada analisis ini mengikuti pengali dari Kaggle.
- Label musim mengikuti dokumentasi dataset: kode 1 = Musim Semi, kode 2 = Musim Panas, kode 3 = Musim Gugur, kode 4 = Musim Dingin. Pada data, Musim Semi (kode 1) mencakup 1 Januari sampai 20 Maret dan 21 sampai 31 Desember, yaitu periode yang secara kalender termasuk musim dingin di Washington D.C., sedangkan Musim Dingin (kode 4) mencakup 23 September sampai 20 Desember. Label musim sebaiknya dibaca sebagai kode kelompok periode pada dataset, bukan musim kalender yang sebenarnya.

### Pertanyaan 1: Pada jam berapa saja tiga jam dengan rata-rata penyewaan sepeda tertinggi pada hari kerja dibandingkan akhir pekan/hari libur sepanjang tahun 2012, dan berapa kali lipat penyewaan jam puncak dibandingkan jam terendah?

```python
h12 = hour_clean[hour_clean["year"] == 2012]
hourly = h12.pivot_table(index="hour", columns="day_type", values="cnt", aggfunc="mean").round(1)

for t in hourly.columns:
    s = hourly[t]
    print(f"== {t} ==")
    print("3 jam tertinggi :", s.nlargest(3).round(0).to_dict())
    print("3 jam terendah  :", s.nsmallest(3).round(0).to_dict())
    print(f"Rasio jam puncak terhadap jam terendah = {s.max() / s.min():.1f} kali\n")
hourly.T
```

**Insight:** (Opsional)
- Hari kerja punya dua puncak komuter: **17.00 (±656), 18.00 (±617), dan 08.00 (±604)**. Rasio jam puncak terhadap jam terendah (03.00) mencapai **119,3 kali**.
- Akhir pekan/libur punya satu puncak lebar di siang hari: **13.00 (±459), 12.00 (±451), dan 14.00 (±448)**. Rasio jam puncak terhadap jam terendah (04.00) sebesar **46,4 kali**.

## Analisis Lanjutan (Opsional) : Manual Grouping / Binning

**Tujuan:** Mengelompokkan data ke dalam kategori berdasarkan aturan bisnis (tanpa algoritma *machine learning*), yaitu:
1. **Kategori suhu** (Dingin <10°C, Sejuk 10–20°C, Hangat 20–28°C, Panas >28°C) untuk melihat ambang suhu yang paling mendorong penyewaan.
2. **Kategori tingkat permintaan per jam** (Rendah, Sedang, Tinggi, Puncak) berdasarkan rata-rata penyewaan per jam, untuk menentukan jam-jam operasional kritis.

```python
# 1) Binning suhu (harian)
day_clean["kategori_suhu"] = pd.cut(day_clean["temp_c"], bins=[-1, 10, 20, 28, 50],
                                    labels=["Dingin (<10°C)", "Sejuk (10–20°C)", "Hangat (20–28°C)", "Panas (>28°C)"])
temp_group = day_clean.groupby("kategori_suhu", observed=True)["cnt"].agg(jumlah_hari="count", rata_rata="mean").round(0)
display(temp_group)

# 2) Binning tingkat permintaan per jam (semua hari)
hour_avg = hour_clean.groupby(["day_type", "hour"])["cnt"].mean().reset_index()
hour_avg["tingkat_permintaan"] = pd.cut(hour_avg["cnt"], bins=[0, 50, 200, 400, 1000],
                                        labels=["Rendah (<50)", "Sedang (50–200)", "Tinggi (200–400)", "Puncak (>400)"])
demand = hour_avg.pivot_table(index="tingkat_permintaan", columns="day_type", values="hour", aggfunc="count", observed=True).fillna(0).astype(int)
display(demand)
```

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
colors = ["#d9d9d9", "#a6a6a6", "#737373", "#404040"]
bars = ax.bar(temp_group.index.astype(str), temp_group["rata_rata"], color=colors)
for b in bars:
    ax.text(b.get_x() + b.get_width() / 2, b.get_height() + 60, f"{b.get_height():,.0f}", ha="center", fontweight="bold")
ax.set_title("Penyewaan meningkat tajam hingga suhu 20°C,\nlalu relatif stabil pada kategori Hangat dan Panas", loc="left", fontsize=12)
ax.set_xlabel("Kategori suhu")
ax.set_ylabel("Rata-rata penyewaan per hari")
ax.set_ylim(0, temp_group["rata_rata"].max() * 1.15)
ax.xaxis.grid(False)
sns.despine()
plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- Rata-rata penyewaan naik tajam dari kategori Dingin (±1.678) ke Sejuk (±3.692), lalu Hangat (±5.581) dan Panas (±5.661). Di atas 20°C kenaikannya melambat dan permintaan cenderung stabil.
- Pada tingkat per jam, jam *Puncak* (>400 sewa) pada hari kerja hanya sedikit (sekitar 3 jam), sehingga persediaan sepeda harus dipusatkan pada jam-jam tersebut.
